```python
# Ejercicio 1 - Verificación de la matriz de transferencia total

s = sp.symbols("s", complex=True)
L1 = sp.symbols("L_1", real=True, positive=True)
L2 = sp.symbols("L_2", real=True, positive=True)
C2 = sp.symbols("C_2", real=True, positive=True)
L3 = sp.symbols("L_3", real=True, positive=True)

Y2 = 1 / (s * L1)
Y3 = 1 / (s * L3)
Zt = (s / C2) / (s**2 + 1 / (C2 * L2))

T1 = sp.Matrix([[1, 1], [0, 1]])
T2 = sp.Matrix([[1, 0], [Y2, 1]])
T3 = sp.Matrix([[1, Zt], [0, 1]])
T4 = sp.Matrix([[1, 0], [Y3, 1]])


Tt = T1 * T2 * T3 * T4

L1_num = sp.Rational(31, 14)
L2_num = sp.Rational(961, 48)
C2_num = sp.Rational(768, 961)
L3_num = sp.Rational(31, 48)


A = Tt[0, 0].simplify()
A = A.subs({L1: L1_num, L2: L2_num, C2: C2_num, L3: L3_num}).simplify()

A_p = (1 + Y2) + Y3 * ((1 + Y2) * Zt + 1).simplify()
A_p_num = A_p.subs({L1: L1_num, L2: L2_num, C2: C2_num, L3: L3_num}).simplify()


markdown_msg = f"""
### Verificación de la matriz de transferencia total\n\n$$ T_t = T_1 \\cdot T_2 \\cdot T_3 \\cdot T_4 $$\n\n

$$ T_1 = {sp.latex(T1)}$$\n\n
$$ T_2 = {sp.latex(T2)}$$\n\n
$$ T_3 = {sp.latex(T3)}$$\n\n
$$ T_4 = {sp.latex(T4)}$$\n\n
$$ T_t = {sp.latex(Tt)}$$

### Se obtiene el parámetro A de la matriz de transferencia total:

$$ A = {sp.latex(A)} $$

### La inversa de A es $\\cfrac{{V_2}}{{V_1}}$ es:

$$ \\frac{{V_2}}{{V_1}} = {sp.latex(1 / A)} $$


### Verificación de la formula de $A$ obtenida a mano:

$$ A = {sp.latex(A_p)} $$

$$ \\frac{{V_2}}{{V_1}} = {sp.latex(1 / A_p_num)} $$
"""

display(Markdown(markdown_msg))
```

```python
# Ejercicio 2 - Verificación de la matriz de transferencia total

C1, C3 = sp.symbols("C_1 C_3", real=True, positive=True)
Y1 = s * C1
Y3 = s * C3
Z2 = (s / C2) / (s**2 + 1 / (C2 * L2))


T1 = sp.Matrix([[1, 0], [Y1, 1]])
T2 = sp.Matrix([[1, Z2], [0, 1]])
T3 = sp.Matrix([[1, 0], [Y3, 1]])
T4 = sp.Matrix([[1, 1], [1, 0]])


C1_num = sp.Rational(27, 17)
C2_num = sp.Rational(27, 289)
L2_num = sp.Rational(289, 243)
C3_num = sp.Rational(7, 17)


Tt = T1 * T2 * T3 * T4

C = Tt[1, 0].simplify()
C = C.subs({C1: C1_num, C2: C2_num, C3: C3_num, L2: L2_num}).simplify()

C_p = Y1 + Y3 * (Y1 * Z2 + 1) + Y1 * Z2 + 1
C_p = C_p.subs({C1: C1_num, C2: C2_num, C3: C3_num, L2: L2_num}).expand().simplify()


markdown_msg = f"""
### Verificación de la matriz de transferencia total\n\n$$ T_t = T_1 \\cdot T_2 \\cdot T_3 \\cdot T_4 $$\n\n

$$ T_1 = {sp.latex(T1)}$$\n\n
$$ T_2 = {sp.latex(T2)}$$\n\n
$$ T_3 = {sp.latex(T3)}$$\n\n
$$ T_4 = {sp.latex(T4)}$$\n\n
$$ T_t = {sp.latex(Tt)}$$

### Se obtiene el parámetro C de la matriz de transferencia total:

$$ C = {sp.latex(C)} $$

### La inversa de C es $\\cfrac{{V_2}}{{I_1}}$ es:

$$ \\frac{{V_2}}{{I_1}} = {sp.latex(1 / C)} $$


### Verificación de la formula de $C$ obtenida a mano:

$$ C = {sp.latex(C_p)} $$

$$ \\frac{{V_2}}{{I_1}} = {sp.latex(1 / C_p)} $$

Nota: Es importante la terminación del circuito. El parámetro C se mide con $I_2 = 0$, lo que implica una carga en cortocircuito al final del circuito.
Por eso el resistor de 1 ohm en el último tramo del circuito se interpreta como si estuviera en serie.
"""

display(Markdown(markdown_msg))
```
